```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
rmin = 100
rmax = 200
img=cv2.imread('/content/lab6.jpg',cv2.IMREAD_GRAYSCALE)
slicing_wo_background = np.zeros(img.shape, dtype=np.uint8)
slicing_w_background = img.copy()

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        pixel_value = img[i, j]
        if rmin <= pixel_value <= rmax:
            slicing_wo_background[i, j] = 255
            slicing_w_background[i, j] = 255

def show_images(*images, titles=None):
    num_images = len(images)
    if titles is None:
        titles = [f'Image {k+1}' for k in range(num_images)]

    plt.figure(figsize=(5 * num_images, 5))
    for k in range(num_images):
        plt.subplot(1, num_images, k + 1)
        plt.imshow(images[k], cmap='gray')
        plt.title(titles[k])
        plt.axis('off')
    plt.show()

show_images(img, slicing_wo_background, slicing_w_background, titles=['Original Image', 'Slicing w/o Background', 'Slicing w Background'])
```
